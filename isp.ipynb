```python
    #   BIBLIOTECAS 
import pandas as pd
import numpy as np 
```

```python
dados = "https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv"
df = pd.read_csv(dados, encoding='latin1', sep=';')
```

```python
df.columns
```

```python
df_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index()
df_cisp = df_cisp.sort_values(by='furto_celular', ascending=False)
df_cisp
```

```python
df_cisp_roubo_veiculo = df.groupby('cisp') ['roubo_veiculo'].sum().reset_index().sort_values(by='roubo_veiculo', ascending=False)
df_cisp_roubo_veiculo
```

```python
df_cisp_roubo_veiculo.head(15)
```

```python
df_cisp_roubo_veiculo.tail(5)
```

```python
df_capital = df[df['munic'] == 'Rio de Janeiro']

# Usando ".loc"
df_capital = df.loc[df['munic'] == 'Rio de Janeiro']

# Usando ".query()"
df_capital = df.query("munic == 'Rio de Janeiro'")

df_cisp_capital = df_capital.groupby('cisp')['furto_celular'].sum().reset_index().sort_values(by='furto_celular', ascending=False)
df_cisp_capital
```

```python
df_filtro_veiculos_rj = df.loc[df['munic'] == 'Rio de Janeiro']

df_furto_veiculos_rj = df_filtro_veiculos_rj.groupby('cisp')['furto_veiculos'].sum().reset_index().sort_values(by='furto_veiculos', ascending=False)

df_furto_veiculos_rj
```

```python
df_furto_veiculos_rj.tail(10)
```

### Estrutura "Array":

```python
array_furto_veiculos_rj = np.array(df_furto_veiculos_rj ['furto_veiculos'])
array_furto_veiculos_rj
```

```python
media = np.mean(array_furto_veiculos_rj)
mediana = np.median(array_furto_veiculos_rj)
q1 = np.percentile(array_furto_veiculos_rj, 25)
q2 = np.percentile(array_furto_veiculos_rj, 50)
q3 = np.percentile(array_furto_veiculos_rj, 75)
print('Q1:', q1)
print('Q2:', q2)
print('Q3:', q3)


print(f'Média {round(media,2)}\nMediana {mediana}') #"round" é usado para arredondar
```

```python
df_top25 = df_furto_veiculos_rj.loc[df_furto_veiculos_rj['furto_veiculos'] > q3]
df_top25
```

### Roubo de celular na Baixada Fluminense:

```python
df_baixada = df.loc[df['regiao'] == 'Baixada Fluminense']

```

```python
df_roubo_baixada = df_baixada.groupby('cisp')['roubo_celular'].sum().reset_index().sort_values(by='roubo_celular', ascending=False)
df_roubo_baixada
```

```python
array_roubo_baixada = np.array(df_roubo_baixada ['roubo_celular'])
array_roubo_baixada
```

```python
mediana_celular = np.median(array_roubo_baixada)
q1 = np.percentile(array_roubo_baixada, 25)
q2 = np.percentile(array_roubo_baixada, 50)
q3 = np.percentile(array_roubo_baixada, 75)
print(f'Mediana: {mediana_celular}')
print('Q1', q1)
print('Q2', q2)
print('Q3', q3)
```

```python
top25_celular = df_roubo_baixada.loc[df_roubo_baixada['roubo_celular'] >q3]
top25_celular
```
