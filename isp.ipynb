```python
import pandas as pd
```

```python
dados = "https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv"
df = pd.read_csv(dados, encoding='latin1', sep=';')
```

```python
df.columns
```

```python
df_cisp = df.groupby('cisp')['furto_celular'].sum().reset_index()
df_cisp = df_cisp.sort_values(by='furto_celular', ascending=False)
df_cisp
```

```python
df_cisp_roubo_veiculo = df.groupby('cisp') ['roubo_veiculo'].sum().reset_index().sort_values(by='roubo_veiculo', ascending=False)
df_cisp_roubo_veiculo
```

```python
df_cisp_roubo_veiculo.tail(5)
```

```python
df_capital = df[df['munic'] == 'Rio de Janeiro']

# Usando ".loc"
df_capital = df.loc[df['munic'] == 'Rio de Janeiro']

# Usando ".query()"
df_capital = df.query("munic == 'Rio de Janeiro'")

df_cisp_capital = df_capital.groupby('cisp')['furto_celular'].sum().reset_index().sort_values(by='furto_celular', ascending=False)
df_cisp_capital
```

```python
df_filtro_veiculos_rj = df.loc[df['munic'] == 'Rio de Janeiro']

df_furto_veiculos_rj = df_filtro_veiculos_rj.groupby('cisp')['furto_veiculos'].sum().reset_index().sort_values(by='furto_veiculos', ascending=False)

df_furto_veiculos_rj
```
