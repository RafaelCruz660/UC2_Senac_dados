### Estrutura "Array":

```python
array_furto_veiculos_rj = np.array(df_furto_veiculos_rj ['furto_veiculos'])
array_furto_veiculos_rj
```

```python
media = np.mean(array_furto_veiculos_rj)
mediana = np.median(array_furto_veiculos_rj)
q1 = np.percentile(array_furto_veiculos_rj, 25)
q2 = np.percentile(array_furto_veiculos_rj, 50)
q3 = np.percentile(array_furto_veiculos_rj, 75)
print('Q1:', q1)
print('Q2:', q2)
print('Q3:', q3)


print(f'Média {round(media,2)}\nMediana {mediana}') #"round" é usado para arredondar
```

### Estelionato:

```python
df_interior = df.loc[df['regiao'] == 'Interior']
df_interior
```

```python
estelionato_municipio = df_interior.groupby('munic')['estelionato'].sum().reset_index().sort_values(by='estelionato', ascending=False)
estelionato_municipio
```

```python
array_estelionato = np.array(estelionato_municipio['estelionato'])
array_estelionato
```

```python
media_estelionato = np.mean(array_estelionato)
mediana_estelionato = np.median(array_estelionato)
distancia = (media_estelionato - mediana_estelionato) / mediana_estelionato
q1 = np.percentile(array_estelionato, 25)
q2 = np.percentile(array_estelionato, 50)
q3 = np.percentile(array_estelionato, 75)
print('Mediana', mediana_estelionato)
print("Q1: ", q1)
print("Q2: ", q2)
print("Q3: ", q3)
print(f"Distancia:  {distancia:.2%}")
```

```python
top25_estelionato = estelionato_municipio.loc[estelionato_municipio['estelionato'] > q3]
top25_estelionato
```

### Medida de tendencia central roubo transeunte


```python
df_transeunte = df.groupby('cisp')['roubo_transeunte'].sum().reset_index().sort_values(by='roubo_transeunte', ascending=False)
df_transeunte
```

```python
array_transeunte = np.array(df_transeunte['roubo_transeunte'])
array_transeunte
```

```python
media_transeunte = np.mean(array_transeunte)
mediana_transeunte = np.median(array_transeunte)
distancia_transeunte = (media_transeunte - mediana_transeunte) / mediana_transeunte
q1 = np.percentile(array_transeunte, 25)
q2 = np.percentile(array_transeunte, 50)
q3 = np.percentile(array_transeunte, 75)
iqr = q3 - q1
limite_superior = q3 + (1.5 * iqr)
limite_inferior = q1 - (1.5 * iqr)

print(f'Media: {media_transeunte:.2f}')
print(f"Mediana: {mediana_transeunte:.2f}")
print(f"Distancia: {distancia_transeunte:.2%}")
print(f"Q1: {q1}")
print(f"Q3: {q3}")
print(f"Limite Superior: {limite_superior}")
print(f"Limite Inferior: {limite_inferior}")
```

```python
outlier_transeunte_cisp = df_transeunte.loc[df_transeunte['roubo_transeunte'] > limite_superior]
outlier_transeunte_cisp
```
