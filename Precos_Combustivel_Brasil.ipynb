```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_202401 = pd.read_csv("AUTOMOTIVOS_2024.01.csv", sep=';', decimal=',')
df_202402 = pd.read_csv("AUTOMOTIVOS_2024.02.csv", sep=';', decimal=',')
df_202501 = pd.read_csv("AUTOMOTIVOS_2025.01.csv", sep=';', decimal=',')
df_202502 = pd.read_csv("AUTOMOTIVOS_2025.02.csv", sep=';', decimal=',')
```

```python
#Concatenando

df = pd.concat([df_202401, df_202402, df_202501, df_202502], ignore_index=True)
df
```

```python
df.columns
```

### Gasolina C no estado

```python
df_rj = df[df['Estado - Sigla'] == 'RJ']
df_rj
```

```python
df_gasolinac_estado = df_rj.groupby('Municipio')['Valor de Venda'].mean().reset_index().sort_values(by='Valor de Venda', ascending=False)
array_gasolinamunicipio = np.array(df_gasolinac_estado['Municipio'])
array_gasolinamunicipiovenda = np.array(df_gasolinac_estado['Valor de Venda'])
df_gasolinac_estado
```

```python
plt.figure(figsize= (20, 10))
plt.bar(array_gasolinamunicipio, array_gasolinamunicipiovenda)
plt.xticks(rotation=90)
plt.title('Gasolina C: \nPreço Medio Por Municipios')
plt.tight_layout()

```

```python
#outliers postos rj municipio

df_gasolinac_postos = df_rj.groupby('Bandeira')['Valor de Venda'].mean().reset_index().sort_values(by='Valor de Venda', ascending=False)
df_gasolinac_postos

```

```python
array_bandeira = np.array(df_gasolinac_postos['Valor de Venda'])
array_bandeiramunic = np.array(df_gasolinac_postos['Bandeira'])
array_bandeiravalor = np.array(df_gasolinac_postos['Valor de Venda'])
```

```python
#Media de tendencia central

media_bandeira = np.mean(array_bandeira)
mediana_bandeira = np.median(array_bandeira)
dist_bandeira = (media_bandeira - mediana_bandeira) / mediana_bandeira
q1 = np.percentile(array_bandeira, 25)
q2 = np.percentile(array_bandeira, 50)
q3 = np.percentile(array_bandeira, 75)
iqr = q3 - q1
limite_superior = q3 + (1.5 * iqr)
limite_inferior = q1 - (1.5 * iqr)

print(f"Media: {media_bandeira}")
print(f"Mediana: {mediana_bandeira}")
print(f"Distancia: {dist_bandeira:.2%}")
print(f"Q1: {q1}")
print(f"Q2: {q2}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
print(f"Limite Superior: {limite_superior}")
print(f"Limite Inferior: {limite_inferior}")
```

```python
plt.figure(figsize= (10, 5))
plt.bar(array_bandeiramunic, array_bandeiravalor)
plt.xticks(rotation=90)
plt.title('Gasolina C: \nGasolina C Por Bandeira')
plt.tight_layout()
```

```python
plt.boxplot(array_bandeira, orientation='horizontal', showmeans=True)
plt.title("Gasolina C Por Bandeira")
```
